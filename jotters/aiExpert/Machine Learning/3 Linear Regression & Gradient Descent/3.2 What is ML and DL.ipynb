## 3.2 What is ML and DL?

আগের page-এ আমরা দেখেছিলাম AI কী এবং কীভাবে এর জন্ম হয়েছিল। কিন্তু AI একটি বড় ছাতার মতো — এর নিচে অনেক কিছু আছে। সেই ছাতার নিচে দুটো সবচেয়ে গুরুত্বপূর্ণ জিনিস হলো Machine Learning এবং Deep Learning। আজকে আমরা বুঝবো এই দুটো আসলে কী, এবং এরা একে অপরের থেকে কীভাবে আলাদা।

#### Traditional Programming vs Machine Learning

প্রথমে একটু পিছনে যাই। পুরনো দিনে programmer-রা যা করতো তা ছিল production rule — একটা problem-এর জন্য manually সব rules লিখে দাও।\
যেমন, cat detect করার জন্য লিখতে হতো:

````python
If (Has Ears) AND (Has Whiskers) AND (Is Furry) THEN → Cat
````

এই approach-এর সমস্যা হলো, real world অনেক complex। যদি cat টা box-এর পেছনে থাকে? যদি whisker ছাড়া কোনো dog থাকে? প্রতিটা exception-এর জন্য নতুন rule লিখতে হবে — এটা কখনো শেষ হবে না।

1959 সালে Arthur Samuel একটা নতুন idea দিলেন। তিনি একটা checkers-playing program বানালেন। তিনি program-কে rules দেননি — দিয়েছিলেন শেখার ক্ষমতা। 

Program নিজে নিজে হাজার হাজার game খেলে, জিতলে কী হয় হারলে কী হয় সেটা দেখে, নিজেই pattem বুঝে নিলো। একসময় সেই program Samuel-কেই হারিয়ে দিলো। এটাই Machine Learning-এর জন্ম।

### Machine Learning কী?
Machine Learning হলো Al-এর এমন একটি approach যেখানে machine-কে explicitly rules দেওয়া হয় না — বরং data দেওয়া হয়, এবং machine সেই data থেকে নিজেই pattern শেখে।

Traditional Programming এবং ML-এর পার্থক্যটা এভাবে দেখা যায়:
Traditional Programming | Machine Learning
---|---
Input | Rules + Data | Data + Answers
Output | Answers | Rules (learned)
কাজ | Programmer rules লেখে | Machine নিজে rules বের করে

ML আজকে কোথায় ব্যবহার হচ্ছে? Gmail-এ spam filter করছে, Netflix recommend করছে, weather predict করছে — সবখানে ML আছে। এই সব জায়গায় কেউ manually rule লেখেনি, machine নিজেই data থেকে শিখেছে।

### Deep Learning কী?
ML-এর মধ্যেই একটি বিশেষ approach আছে যেটা মানুষের brain থেকে inspired — সেটা হলো Deep Learning। মানুষের brain-এ কোটি কোটি neuron আছে, এরা interconnected। 

একটা neuron অন্য neuron থেকে signal নেয়, process করে, পাঠায়। Deep Learning-এ এই structure-টাকেই artificially তৈরি করা হয় — এটাকে বলে Neural Network।

একটা Neural Network-এ তিনটি অংশ থাকে:

```mermaid
flowchart LR
    A["Input Layer<br/>(Raw Data)"] --> B["Hidden Layers<br/>(Pattern Finding)"]
    B --> C["Output Layer<br/>(Answer)"]
```
উদাহরণ হিসেবে একটা cat photo detect করার কথা ভাবো। Input layer-এ ছবির প্রতিটি pixel আসে। Hidden layers গুলো প্রথমে edges detect করে, তারপর সেই edges মিলিয়ে shapes বানায়, তারপর shapes মিলিয়ে features (যেমন ear, whisker) বানায়। শেষে Output layer বলে — "এটা Cat"

"Deep" মানে হলো এই Hidden Layers অনেক বেশি layer যত বেশি, network তত "deep"।

### তিনটির সম্পর্ক
এই তিনটি concept অনেকে গুলিয়ে ফেলে। আসলে এরা একটার মধ্যে আরেকটা:

```mermaid
flowchart TB
    subgraph AI["Artificial Intelligence"]
        subgraph ML["Machine Learning"]
            DL["Deep Learning"]
        end
    end
```

Al হলো সবচেয়ে বড় concept — machine দিয়ে intelligent কাজ করানো। ML হলো AI করার একটি পদ্ধতি — data থেকে শেখা। Deep Learning হলো ML-এর একটি বিশেষ technique — Neural Network ব্যবহার করে শেখা।

সব ML, Deep Learning না। কিন্তু সব Deep Learning, ML-এর অংশ।

### কোনটা কখন ব্যবহার হয়?
Situation | কোনটা ব্যবহার হয়
---|---
Structured data, simple pattern | Traditional ML
Images, Audio, Text | Deep Learning
খুব কম data | Traditional ML
বিশাল data আছে | Deep Learning

Deep Learning অনেক powerful, কিন্তু এর জন্য দরকার প্রচুর data এবং powerful hardwarel Traditional ML কম data-তেও ভালো কাজ করে।

### Summary
Traditional programming-9 manually rules লিখতে হতো, Machine Learning-& machine নিজেই data থেকে rules শেখে। Deep Learning হলো ML-এর একটি বিশেষ approach যেখানে human brain-এর মতো Neural Network ব্যবহার করা হয়। AI > ML > Deep Learning — এই তিনটি একটার মধ্যে আরেকটা nested, আলাদা আলাদা জিনিস না।
