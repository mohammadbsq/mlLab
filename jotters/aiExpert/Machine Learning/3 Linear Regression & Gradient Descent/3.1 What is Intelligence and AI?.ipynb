## 3.1 What is Intelligence and Al?
মানুষ হাজার বছর ধরে একটা স্বপ্ন দেখে এসেছে &mdash; এমন কিছু তৈরি করা যা তার মতোই চিন্তা করতে পারে, শিখতে পারে, সিদ্ধান্ত নিতে পারে। 

এই স্বপ্নই আজকের Artificial Intelligence-এর জন্ম দিয়েছে। কিন্তু machine কে intelligent করতে হলে আগে বুঝতে হবে &mdash; intelligence আসলে কী?

### Intelligence কী?
Intelligence মানে শুধু facts জানা না। Intelligence হলো একটা cyclic process &mdash; পরিবেশ থেকে তথ্য নাও, সেটা থেকে শেখো, সিদ্ধান্ত নাও, কাজ করো, আবার নতুন তথ্য নাও। এই loop-টাই intelligent behavior-এর মূল।

একটা শিশু যখন হাঁটতে শেখে, সে প্রতিবার পড়ে যায়, তারপর adjust করে, আবার চেষ্টা করে। একজন ডাক্তার যখন রোগ নির্ণয় করেন, তিনি past experience থেকে pattern match করেন। এই ability গুলোকে চারটি ভাগে ভাগ করা যায়:
- Learn: নতুন তথ্য absorb করা
- Adapt: নতুন পরিবেশে টিকে থাকা
- Reason: জটিল সমস্যা সমাধান করা
- Remember: অতীত ব্যবহার করে ভবিষ্যৎ guide করা

মানুষ এই চারটি কাজ স্বাভাবিকভাবেই করতে পারে। প্রশ্ন হলো &mdash; machine কি পারবে?

### AI-এর জন্ম: Dartmouth, 1956
1955 সালে McCarthy, Minsky, Shannon এবং Rochester মিলে একটা proposal লিখলেন। তাঁদের ধারণা ছিল intelligence-এর প্রতিটি দিক এত precisely describe করা সম্ভব যে machine সেটা simulate করতে পারবে। 

এই proposal-এর ভিত্তিতে 1956 সালে Dartmouth College-এ একটি summer workshop হয় এবং সেখানেই প্রথমবার "Artificial Intelligence" শব্দটি officially ব্যবহার হয়।

 তার আগে 1950 সালে Alan Turing একটি প্রশ্ন রেখেছিলেন- "Can machines think?" তিনি বলেছিলেন, যদি একটি machine কোনো মানুষের সাথে conversation করে এবং সেই মানুষ বুঝতে না পারে যে সে machine-এর সাথে কথা বলছে, তাহলে সেই machine কে intelligent বলা যায়। এটাই Turing Test।
 
### AI আসলে কী?
সহজ ভাষায়, AI হলো এমন software যা এমন কাজ করতে পারে যেগুলো সাধারণত human intelligence দরকার হয়- যেমন দেখা, শোনা, কথা বলা, সিদ্ধান্ত নেওয়া।

কিন্তু আজকের AI এবং মানুষের intelligence-এর মধ্যে একটা বড় পার্থক্য আছে। এখন দুটো ভাগ আছে:

| | ANI (Narrow AI) | AGI (General AI) |
|---|---|---|
| **সংজ্ঞা** | একটি নির্দিষ্ট কাজে expert | যেকোনো কাজে সক্ষম |
| **উদাহরণ** | Chess AI, Face Recognition | Human-level thinking |
| **বর্তমান অবস্থা** | বিদ্যমান | এখনও hypothetical |


আজকের সব Al - ChatGPT, Google Photos, Siri - সবই ANI। এরা প্রত্যেকে একটা নির্দিষ্ট কাজে অসাধারণ, কিন্তু সেই কাজের বাইরে কিছু করতে পারে না। Chess খেলা Al কোনো poem লিখতে পারবে না।
কেন AI এখন এত শক্তিশালী?
AI 1950 সাল থেকেই আছে। তাহলে হঠাৎ এখন এত বড় হলো কেন? তিনটি কারণ একসাথে মিলেছে:
প্রথমত, Big Data - internet-এর কারণে এখন machine কে শেখানোর জন্য কোটি কোটি example পাওয়া যায়। দ্বিতীয়ত,
Powerful GPUs - এমন hardware তৈরি হয়েছে যা এই বিশাল data দ্রুত process করতে পারে। তৃতীয়ত, Better
Algorithms - গত দশ বছরে mathematics এবং optimization-এ নতুন discoveries হয়েছে।
এই তিনটি জিনিস একসাথে মিলেই Al revolution সম্ভব হয়েছে।
