## 1.13 Y Data Profiling
এখন পর্যন্ত আমরা আলাদা আলাদা plot বানিয়ে data analyze করেছি। প্রতিটি plot-এর জন্য আলাদা code লিখতে হয়েছে।
কিন্তু একটি নতুন dataset পেলে প্রথমেই কী করা উচিত? কতগুলো column আছে, কোথায় missing values আছে, কোন column-এর distribution কেমন এই সব একসাথে দেখতে অনেক সময় লাগে। Y Data Profiling এই কাজটি মাত্র কয়েক
line code-এ করে দেয়।

### What is Y Data Profiling?
Y Data Profiling (আগের নাম ছিল Pandas Profiling) একটি Python library যা একটি DataFrame নিয়ে automatically
একটি complete EDA report তৈরি করে। এই report-এ থাকে:
- প্রতিটি column-এর data type এবং basic statistics (mean, median, min, max)
- Missing values কোথায় কতটুকু আছে
- প্রতিটি numerical column-এর distribution
- Columns-এর মধ্যে correlation
- Duplicate rows আছে কিনা
- Outliers-এর সংখ্যা


### Generating the Report
Install হয়ে গেলে মাত্র কয়েক line code-এ report তৈরি করা যায়:


```python
import pandas as pd
from ydata_profiling import ProfileReport

df = pd.read_csv('titanic_data_updated.csv')

profile = ProfileReport(df, title='Titanic Profile Report')

profile.to_file('my_report.html')
```

এখানে:
- ProfileReport(df, title=...) DataFrame নিয়ে report object তৈরি করে
- title=... report-এর উপরে যে নাম দেখাবে
- profile.to_file('my_report.html') → report-টি HTML file হিসেবে save করে

- Reading the Report
Report generate হলে my_report.html নামে একটি file তৈরি হবে। এই file browser-এ open করলে একটি interactive report দেখা যাবে। Report-এ কয়েকটি section থাকে:

| Section | কী দেখায় |
|---|---|
| Overview | Dataset-এর মোট rows, columns, missing values-এর summary |
| Variables | প্রতিটি column-এর আলাদা analysis distribution, statistics |
| Interactions | দুটো numerical column-এর মধ্যে scatterplot |
| Correlations | সব columns-এর মধ্যে correlation heatmap |
| Missing Values | কোন column-এ কতটা data missing |
| Sample | Dataset-এর প্রথম এবং শেষ কয়েকটি row |

### Y Data Profiling vs Manual EDA
| বিষয় | Manual EDA | Y Data Profiling |
|---|---:|:---|
| সময় | অনেক বেশি — প্রতিটি plot আলাদা করে বানাতে হয় | কয়েক সেকেন্ড |
| কোড | অনেক line | মাত্র ৩-৪ line |
| Coverage | যা plot করবো শুধু তাই দেখাবে | সব columns automatically analyze হয় |
| কখন use | নির্দিষ্ট analysis দরকার হলে | নতুন dataset পেলে প্রথমেই overview নিতে করবো |

### When to Use Y Data Profiling?
Y Data Profiling সবচেয়ে বেশি কাজে আসে যখন:
নতুন একটি dataset পেয়েছি এবং প্রথমে একটি quick overview দরকার
Dataset-এ কোথায় missing values আছে দ্রুত জানতে চাই
সব columns-এর distribution একসাথে দেখতে চাই
Client বা teammate-কে data-র summary দেখাতে চাই
