### 1.11 KDE Plot and Density Graph
আমরা এর আগে Histogram দেখেছি যেটি একটি numerical variable-এর distribution বোঝার জন্য ব্যবহার হয়। Histogram-এ data-কে ছোট ছোট bins-এ ভাগ করে bar দিয়ে দেখানো হয়। কিন্তু Histogram-এর একটি সমস্যা আছে - bin size পরিবর্তন করলে distribution-এর shape পরিবর্তন হয়ে যায়। এই সমস্যার সমাধান হলো KDE Ploti
### What is KDE Plot?
KDE মানে Kernel Density Estimation। এটি Histogram-এর মতোই একটি numerical variable-এর distribution দেখায়, কিন্তু bars-এর বদলে একটি smooth continuous curve তৈরি করে। এই curve দেখে বোঝা যায় data কোন range-এ বেশি concentrated এবং কোথায় কম।

সহজ উদাহরণ দিয়ে বলতে গেলে ধরো একটি ক্লাসের সব ছাত্রের বয়স plot করলে। Histogram দেখাবে প্রতিটি age range-এ কতজন আছে bar দিয়ে। KDE plot দেখাবে সেই একই information কিন্তু একটি smooth wave-এর মতো curve দিয়ে - যা পড়তে অনেক সহজ।
### KDE Plot with Hue
KDE plot-এর সবচেয়ে শক্তিশালী ব্যবহার হলো `hue` parameter দিয়ে। এতে আলাদা আলাদা category-র জন্য আলাদা।\
KDE curve দেখা যায় এবং সেগুলো একসাথে compare করা যায়।
Titanic data-তে Age-এর distribution survive করা এবং না করা passengers-এর মধ্যে কেমন আলাদা সেটা দেখি:


```python
sns.kdeplot(data=df, x='Age', hue='Survived')

plt.title('Age Distribution by Survival')
plt.show()
```


এই plot থেকে দেখা যায়:
যারা survive করেছে তাদের মধ্যে শিশু (ছোট বয়স) বেশি ছিল
যারা survive করেনি তাদের মধ্যে মধ্যবয়সী passengers বেশি ছিল
দুটো group-এর Age distribution-এর shape আলাদা
এই insight শুধু Histogram দিয়ে এত সহজে বোঝা যেত না।

### KDE vs Histogram
| | Histogram | KDE Plot |
|---|---|---|
| Output | Discrete bars | Smooth continuous curve |
| Bin dependency | হ্যাঁ - bin size বদলালে shape বদলায় | নেই |
| Multiple groups compare | কঠিন | সহজ - hue দিয়ে overlay করা যায় |
| Y-axis | Count বা frequency | Density (0 থেকে। এর মধ্যে) |
| কখন use করবো | Exact count দেখতে | Distribution-এর shape বুঝতে |

Understanding the Y-axis\
KDE plot-এর y-axis density দেখায়, count নয়। Density মানে হলো- curve-এর নিচের মোট area সবসময় 1 হয়। যেখানে curve উঁচু, সেখানে data বেশি concentrated। যেখানে curve নিচু, সেখানে data কম।\
এটা একটু confusing মনে হতে পারে শুরুতে। সহজ করে মনে রাখো - curve যত উঁচু, সেই range-এ তত বেশি data আছে।
### When to Use KDE Plot?
KDE plot তখন use করবো যখন:
- একটি continuous numerical variable-এর distribution-এর shape দেখতে চাই
- দুই বা তার বেশি group-এর distribution compare করতে চাই
- Data কোন range-এ বেশি concentrated সেটা smooth curve দিয়ে বুঝতে চাই
### Summary

parameter যোগ করলে এটি Multivariate Analysis-এর একটি শক্তিশালী tool হয়ে ওঠে। Titanic data-তে আমরা দেখলাম survive করা এবং না করা passengers-এর Age distribution কতটা আলাদা যা Barplot বা Countplot দিয়ে এতটা clearly বোঝানো সম্ভব হতো না।
