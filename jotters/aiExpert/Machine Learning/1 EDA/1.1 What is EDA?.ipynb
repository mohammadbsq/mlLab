## 1.1 What is EDA?

EDA এর পূর্ণরূপ হলো **Exploratory Data Analysis**।\
বাংলায় একে বলা যায় **ডেটা অনুসন্ধানমূলক বিশ্লেষণ**।

Machine Learning বা Data Science প্রজেক্টে Model তৈরির আগে Dataset সম্পর্কে ভালোভাবে বোঝার জন্য যে বিশ্লেষণ করা হয়, সেটাই EDA।

সহজভাবে বললে, EDA হলো Dataset এর ভিতরে কী আছে, কোন ধরনের তথ্য আছে, কোথায় সমস্যা আছে, কোন Pattern আছে - এসব খুঁজে বের করার প্রক্রিয়া।

**EDA** is a systematic process of investigating datasets to understand their structure, patterns, and quality before building machine learning models.

### Why EDA Matters?
অনেক সময় Dataset দেখতে ঠিক মনে হলেও ভিতরে বিভিন্ন সমস্যা থাকতে পারে। যেমন:

- Missing Value থাকতে পারে
- ভুল Data Type থাকতে পারে
- Duplicate Data থাকতে পারে
- Outlier থাকতে পারে
- Irrelevant Feature থাকতে পারে
- Data Imbalance থাকতে পারে

যদি এসব সমস্যা আগে থেকে শনাক্ত না করা হয়, তাহলে Machine Learning Model ভালো Result দিতে পারবে না।\
তাই EDA হলো Data Science Workflow এর সবচেয়ে গুরুত্বপূর্ণ ধাপগুলোর একটি।

There is an AI principle known as \
**"Garbage In, Garbage Out" (GIGO)**

Poor quality data leads to unreliable models. EDA helps ensure data quality and reliability before proceeding to model development.

### Key Objectives:

EDA করার সময় সাধারণত নিচের বিষয়গুলো বোঝার চেষ্টা করা হয়:

**1. Dataset Structure বোঝা**
- কতগুলো Row আছে
- কতগুলো Column আছে
- কোন Column Numerical
- কোন Column Categorical

**2. Data Quality যাচাই করা**

- Missing Value আছে কি না
- Duplicate আছে কি না
- Invalid or Outlier data আছে কি না

**3. Statistical Understanding তৈরি করা**

- Mean
- Median
- Mode
- Standard Deviation
- Minimum & Maximum Value

**4. Pattern এবং Relationship খুঁজে বের করা**

- কোন Feature গুরুত্বপূর্ণ
- কোন Feature Target এর সাথে সম্পর্কিত
- কোন Feature অপ্রয়োজনীয়

**5. Visualization এর মাধ্যমে Data বোঝা**

Graph ব্যবহার করে Data আরও সহজে বিশ্লেষণ করা যায়। 

যেমন:

- Histogram
- Boxplot
- Countplot
- Scatterplot


### Real Life Example

ধরুন আপনি একটি Student Dataset নিয়ে কাজ করছেন।

Dataset এ নিচের তথ্যগুলো আছে:

| Name   | Age | Gender | CGPA |
|--------|-----|--------|------|
| Rahim  | 22  | Male   | 3.75 |
| Karima | 21  | Female | 3.90 |

এখন Model তৈরির আগে আপনি জানতে চাইবেন:
- Average CGPA কত?
- Male/Female ratio কত?
- কোনো Missing Value আছে কি?
- Age এবং CGPA এর মধ্যে সম্পর্ক আছে কি?

এই পুরো Investigation Process-ই হলো EDA।

### EDA Process:

সাধারণত EDA কয়েকটি ধাপে করা হয়:
1. Data Understanding
2. Data Cleaning
3. Feature Screening
4. Statistical Profiling
5. Univariate Analysis
6. Bivariate Analysis
7. Multivariate Analysis
8. Data Visualization


### EDA Types:
EDA সাধারণত দুই ধরনের Analysis নিয়ে কাজ করে:

**1. Univariate Analysis**
একটি Column আলাদাভাবে বিশ্লেষণ করা।

উদাহরণ:
- Age distribution
- Gender count

**2. Multivariate Analysis**
একাধিক Column এর মধ্যে সম্পর্ক বিশ্লেষণ করা।

উদাহরণ:
- Age vs Salary
- Study Hour vs CGPA


### EDA & Machine Learning:
EDA ছাড়া ভালো Machine Learning Model তৈরি করা কঠিন।

কারণ EDA:
- Data সম্পর্কে ধারণা দেয়
- Feature Selection সহজ করে
- Data Cleaning করতে সাহায্য করে
- Better Model Performance নিশ্চিত করে

### Important Things:
EDA শুধুমাত্র Graph আঁকার কাজ নয়।\
অনেকেই মনে করে EDA মানেই Visualization।

কিন্তু বাস্তবে EDA হলো:
- Data বোঝা
- Data যাচাই করা
- Statistical Analysis করা
- Pattern খুঁজে বের করা
- Business Insight বের করা

Visualization হচ্ছে EDA এর একটি অংশ মাত্র।

### Summary:
EDA হলো Dataset কে গভীরভাবে বোঝার প্রক্রিয়া।

এর মাধ্যমে আমরা:
- Data Structure বুঝি
- Problem খুঁজে বের করি
- Important Feature শনাক্ত করি
- Pattern আবিষ্কার করি
- Model তৈরির জন্য Data প্রস্তুত করি

তাই বলা যায়:

"EDA is the process of understanding data before building any Machine Learning model."
