## 1.12 Scatterplot
এখন পর্যন্ত আমরা যে plots শিখেছি Countplot, Barplot, KDE Plot - সেগুলো মূলত categorical variable-কে কেন্দ্র করে। কিন্তু যখন দুটো numerical variable-এর মধ্যে সম্পর্ক দেখতে হয়, তখন Scatterplot সবচেয়ে কার্যকর tool।
Scatterplot-এ প্রতিটি data point একটি dot হিসেবে দেখা যায় এবং dots-এর pattern দেখে দুটো variable-এর মধ্যে সম্পর্ক বোঝা যায়।

### What is a Scatterplot?
Scatterplot-9 x-axis-এ একটি numerical variable এবং y-axis-এ আরেকটি numerical variable রাখা হয়। Dataset-এর প্রতিটি row একটি dot হিসেবে plot-এ দেখা যায়। dots-এর overall pattern দেখে বোঝা যায় দুটো variable একসাথে বাড়ে কিনা, একটি বাড়লে অন্যটি কমে কিনা, বা কোনো সম্পর্কই নেই।

এই page-এ আমরা Titanic-এর বদলে Seaborn-এর built-in tips dataset ব্যবহার করবো। এই dataset-এ একটি
restaurant-এর customer-দের bill এবং tip-এর data আছে।



```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')
df
```

Dataset-এ আছে: total_hill, tip, sex, smoker, day, time, size এই columns


Basic Scatterplot


```python
sns.scatterplot(data=df,
                x='total_bill',
                y='tip')

plt.title('Total Bill vs Tip')
plt.show()
```


এই plot থেকে দেখা যায় total bill বাড়লে সাধারণত tip-ও বাড়ে দুটো variable-এর মধ্যে একটি positive সম্পর্ক আছে।

### Scatterplot with Hue and Size
Scatterplot-9 hue এবং size দেখানো সম্ভব হয়। দুটো extra parameter যোগ করা যায়। এতে একটি plot-এই চারটি variable একসাথে দেখানো সম্ভব হয়।


```python
sns.scatterplot(data=df,
                x='total_bill',
                y='tip',
                hue='sex',
                size='size')

plt.title('Total Bill vs Tip by Sex and Party Size')
plt.show()
```


এখানে:
- x-axis total_bill (numerical)
- y-axis tip (numerical)
- hue sex (categorical) - male 3 female আলাদা color-এ দেখাবে
- size size (numerical) - party-তে কতজন ছিল, সেই অনুযায়ী dot-এর size বড়-ছোট হবে

এই একটি plot থেকে আমরা একসাথে জানতে পারি bill এবং tip-এর সম্পর্ক, male ও female-এর pattern আলাদা কিনা, এবং বড় party-তে tip বেশি দেওয়া হয় কিনা।

### Reading a Scatterplot
Pattern | মানে
---|---
Dots উপরের দিকে যাচ্ছে | Positive relationship - একটি বাড়লে অন্যটিও বাড়ে
Dots নিচের দিকে যাচ্ছে | Negative relationship - একটি বাড়লে অন্যটি কমে
Dots randomly ছড়ানো | কোনো সম্পর্ক নেই
Dots একটি curve তৈরি করছে | Non-linear relationship আছে

When to Use Scatterplot?
Scatterplot তখন use করবো যখন:
- দুটো numerical variable-এর মধ্যে সম্পর্ক আছে কিনা দেখতে চাই
- সেই সম্পর্ক কোনো categorical variable-এর জন্য আলাদা কিনা দেখতে চাই
- Data-তে outlier আছে কিনা visually check করতে চাই

### Summary
Scatterplot দুটো numerical variable-এর মধ্যে সম্পর্ক dot দিয়ে দেখায়। hue এবং size parameter যোগ করলে এটি একটি শক্তিশালী Multivariate Analysis tool হয়ে ওঠে। Tips dataset-এ আমরা দেখলাম total bill এবং tip-এর মধ্যে positive সম্পর্ক আছে এবং hue=sex ও size=size যোগ করে আরো গভীর insight পাওয়া গেল।
