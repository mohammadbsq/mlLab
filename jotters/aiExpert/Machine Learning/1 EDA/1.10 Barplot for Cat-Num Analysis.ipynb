## 1.10 Barplot for Cat-Num Analysis
আমরা আগের page-এ দেখলাম Countplot with Hue দিয়ে দুটো categorical variable-এর মধ্যে সম্পর্ক বোঝা যায়। কিন্তু যখন একটি categorical variable এবং একটি numerical variable-এর মধ্যে সম্পর্ক দেখতে হয়, তখন Barplot ব্যবহার করা হয়। Cat-Num মানে Categorical-Numerical - এই ধরনের analysis-এ Barplot সবচেয়ে কার্যকর tooll

### What Does Barplot Show?
Seaborn-এর Barplot প্রতিটি categorical group-এর জন্য একটি numerical variable-এর mean (গড়) দেখায়। প্রতিটি bar-এর উপরে একটি ছোট line থাকে যাকে confidence interval বলে এটি বলে দেয় mean-এর আশেপাশে data কতটা spread করা আছে।

### Example 1: Pclass vs Fare


```python


sns.barplot(x=df['Pclass'], y=df['Fare'])

plt.title('Average Fare by Passenger Class')
plt.show()
```

এই plot থেকে দেখা যায় first class passengers গড়ে সবচেয়ে বেশি Fare দিয়েছে এবং third class সবচেয়ে কম। x-axis-A Pclass (categorical) এবং y-axis-এ Fare (numerical)- এটাই Cat-Num analysis-এর basic structurel
### Example 2: Survived vs Age with Hue
Barplot-এও `hue` ব্যবহার করা যায়। এতে তিনটি variable একসাথে দেখা যায় একটি categorical (x-axis), একটি numerical (y-axis), এবং আরেকটি categorical (hue) 


```python
sns.barplot(x=df['Survived'], y=df['Age'], hue=df['Sex'])

plt.title('Average Age of Survivors by Sex')
plt.show()
```


এই plot থেকে বোঝা যায়- যারা survive করেছে এবং যারা করেনি, তাদের গড় Age কত ছিল এবং এই pattern male 3 female-এর মধ্যে কেমন আলাদা।
hue ছাড়া শুধু Survived vs Age দেখলে:

```python
sns.barplot(x=df['Survived'], y=df['Age'])

plt.title('Average Age by Survival')
plt.show()
```

এই দুটো plot compare করলে বোঝা যায় hue যোগ করলে কতটা বেশি information পাওয়া যায়।
### Barplot vs Countplot

| | Countplot | Barplot |
|---|---|---|
| x-axis | Categorical | Categorical |
| y-axis | Count (automatic) | Numerical (manual) |
| দেখায় | Frequency | Mean of numerical variable |
| hue support | ✓ | ✓ |
| কখন use করবো | Category কতবার আছে জানতে | Category অনুযায়ী numerical value-এর গড় জানতে |

When to Use Barplot?

Barplot তখন use করবো যখন:
- জানতে চাই বিভিন্ন group-এর numerical variable-এর গড় কত
- যেমন- প্রতিটি Pclass-এর গড় Fare কত?
- যেমন - survive করা এবং না করা passengers-এর গড় Age কত?

### Summary

Barplot Cat-Num analysis-এর সবচেয়ে সহজ এবং কার্যকর tool। এটি categorical group অনুযায়ী numerical variable-এর mean দেখায়। `hue` যোগ করলে একসাথে তিনটি variable visualize করা যায় যা data-র আরো গভীর pattern বুঝতে সাহায্য করে। Titanic data-তে আমরা দেখলাম Pclass, Fare, Survived এবং Age - এই চারটি variable একসাথে Barplot দিয়ে analyze করা সম্ভব হলো।
