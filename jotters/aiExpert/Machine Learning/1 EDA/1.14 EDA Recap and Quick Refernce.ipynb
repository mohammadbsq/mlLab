## 1.14 EDA Recap and Quick Refernce.ipynb
আমরা Multivariate Analysis-এর সবচেয়ে গুরুত্বপূর্ণ plots এবং tools শিখেছি। প্রতিটি tool আলাদা পরিস্থিতিতে কাজে আসে। এই page-এ সব কিছু একসাথে recap করবো এবং একটি quick reference তৈরি করবো যাতে কোন situation-এ কোন plot use করতে হবে সেটা সহজে মনে রাখা যায়।
### What We Learned
Countplot with Hue দিয়ে আমরা দেখলাম দুটো categorical variable-এর মধ্যে সম্পর্ক। Titanic data-তে Sex এবং
Pclass - উভয়ই Survival-এর সাথে কতটা সম্পর্কিত সেটা color দিয়ে আলাদা করে বোঝা গেল।
Barplot দিয়ে আমরা দেখলাম categorical group অনুযায়ী numerical variable-এর meani Pclass অনুযায়ী গড় Fare কত এবং Survived অনুযায়ী গড় Age কত এই ধরনের প্রশ্নের উত্তর Barplot সহজে দেয়।
KDE Plot দিয়ে আমরা দেখলাম continuous variable-এর smooth distributioni hue যোগ করে survive করা এবং না করা passengers-এর Age distribution compare করলাম এবং দুটো group-এর মধ্যে স্পষ্ট পার্থকা দেখতে পেলাম।
Scatterplot দিয়ে আমরা দেখলাম দুটো numerical variable-এর মধ্যে সম্পর্ক। Tips dataset-& total bill এবং tip-এর মধে positive relationship ছিল এবং hue ও size যোগ করে আরো গভীর insight পেলাম।
Y Data Profiling দিয়ে আমরা দেখলাম কীভাবে মাত্র কয়েক line code-এ পুরো dataset-এর automatic EDA report তৈরি করা যায়।

### Quick Reference: Which Plot When?

| পরিস্থিতি | কোন Plot | Example |
|---|---|---|
| দুটো categorical variable compare করতে চাই | Countplot with Hue | Sex vs Survived |
| Categorical group-এর numerical mean দেখতে চাই | Barplot | Pclass vs Fare |
| Categorical group-এর numerical mean তৃতীয় variable সহ দেখতে চাই | Barplot with Hue | Survived vs Age, hue=Sex |
| Continuous variable-এর distribution দেখতে চাই | KDE Plot | Age distribution |
| দুটো group-এর distribution compare করতে চাই | KDE Plot with Hue | Age by Survived |
| দুটো numerical variable-এর সম্পর্ক দেখতে চাই | Scatterplot | total_bill vs tip |
| নতুন dataset-এর quick overview নিতে চাই | Y Data Profiling | Titanic full report |

### Variable Type Guide
কোন ধরনের data-তে কোন plot কাজ করে:
| পরিস্থিতি | কোন Plot | Example |
|---|---|---|
| দুইটি categorical variable compare করতে চাই | Countplot with Hue | Sex vs Survived |
| Categorical group-এর numerical mean দেখতে চাই | Barplot | Pclass vs Fare |
| Categorical group-এর numerical mean তৃতীয় variable সহ দেখতে চাই | Barplot with Hue | Survived vs Age, hue=Sex |
| Continuous variable-এর distribution দেখতে চাই | KDE Plot | Age distribution |
| দুইটি group-এর distribution compare করতে চাই | KDE Plot with Hue | Age by Survived |
| দুইটি numerical variable-এর সম্পর্ক দেখতে চাই | Scatterplot | total_bill vs tip |
| নতুন dataset-এর quick overview নিতে চাই | Y Data Profiling | Titanic full report |

### Key Takeaways
Multivariate Analysis-এর মূল কথা হলো- data-র পূর্ণ চিত্র দেখতে হলে একসাথে তিন বা তার বেশি variable analyze করতে হয়। hue parameter প্রায় সব Seaborn plot-এ কাজ করে এবং এটি দিয়ে যেকোনো Bivariate analysis-কে Multivariate-এ নিয়ে যাওয়া যায়।
নতুন dataset পেলে সবার আগে Y Data Profiling দিয়ে overview নাও, তারপর specific questions-এর জন্য আলাদা আলাদা plot বানাও।
