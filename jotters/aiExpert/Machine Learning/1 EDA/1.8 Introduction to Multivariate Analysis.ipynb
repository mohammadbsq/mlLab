## 1.8 Introduction to Multivariate Analysis
আমরা এর আগে Univariate এবং Bivariate Analysis শিখেছি। Univariate Analysis-এ আমরা একটিমাত্র variable নিয়ে কাজ করতাম - যেমন শুধু Age-এর distribution দেখা। Bivariate Analysis-এ দুটো variable-এর মধ্যে সম্পর্ক দেখতাম যেমন Age এবং Fare। কিন্তু real-world data-তে একটি outcome সাধারণত একটি বা দুটি variable-এর উপর নির্ভর করে না।
একসাথে অনেক variable মিলে একটি pattern তৈরি করে। এই জায়গাতেই Multivariate Analysis আসে।
### What is Multivariate Analysis?
Multivariate Analysis হলো একসাথে তিন বা তার বেশি variable-এর মধ্যে সম্পর্ক বোঝার পদ্ধতি। এটি আমাদের বলে - শুধু "Age বেশি হলে কী হয়" নয়, বরং "Age বেশি হলে এবং Sex female হলে Survival rate কেমন হয়?" এই ধরনের প্রশ্নের উত্তর Multivariate Analysis দেয়।
উদাহরণ হিসেবে Titanic dataset-এর কথা ভাবো। একজন passenger survive করবে কিনা সেটা নির্ভর করে তার Sex, Pclass, Age, Fare - একসাথে অনেক কিছুর উপর। শুধু একটা variable দেখে সিদ্ধান্ত নিলে আমরা অনেক গুরুত্বপূর্ণ pattern miss করবো।
### Univariate vs Bivariate vs Multivariate
| Type | Variables | Example | Plot |
|------|-----------|---------|------|
| Univariate | ১টি | Age-এর distribution | Histogram, KDE |
| Bivariate | ২টি | Age vs Fare | Scatterplot, Barplot |
| Multivariate | ৩+ টি | Age, Sex, Survived একসাথে | Countplot with Hue, KDE with Hue |
### Why Do We Need It?
শুধু Bivariate Analysis দিয়ে আমরা দেখতে পারি Pclass এবং Survived-এর মধ্যে সম্পর্ক আছে। কিন্তু Multivariate Analysis দিয়ে আমরা দেখতে পারি first class-এর female passengers-এর survival rate কত, এবং third class-এর male passengers-এর survival rate কত। এই deeper insight machine learning model বানাতে অনেক কাজে আসে। সহজ করে বলতে গেলে Univariate Analysis বলে "কী আছে", Bivariate Analysis বলে "দুটোর মধ্যে সম্পর্ক কী", আর Multivariate Analysis বলে "আসল ছবিটা কেমন।"

### What We Will Learn
এই module-এ আমরা Multivariate Analysis-এর জন্য সবচেয়ে বেশি ব্যবহৃত plots শিখবো।

Scatterplot - দুটো numerical variable-এর relationship que এবং size দিয়ে আরো গভীরভাবে দেখা
Y Data Profiling একটি command-এ পুরো dataset-এর automatic analysis
### Summary
Multivariate Analysis আমাদের data-র পূর্ণ চিত্র দেখতে সাহায্য করে। একটি বা দুটি variable দিয়ে যে pattern ধরা যায় না, তিনটি বা তার বেশি variable একসাথে দেখলে সেই pattern স্পষ্ট হয়ে ওঠে। এই module জুড়ে আমরা real Titanic data ব্যবহার করে প্রতিটি plot হাতে-কলমে শিখবো।
