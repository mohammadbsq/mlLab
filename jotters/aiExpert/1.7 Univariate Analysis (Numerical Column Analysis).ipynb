### Numerical Analysis কেন গুরুত্বপূর্ণ?
Numerical Analysis এর মাধ্যমে আমরা বুঝতে পারি:
- Data Distribution কেমন

### Age Column এর Histogram
প্রথমে Age Column এর Histogram তৈরি করা যাক।

```python
sns.histplot(df['Age'],bins=50)

plt.title('Histogram of Age')
plt.xlabel('Ages')
plt.ylabel('frequency')

plt.show()
```

### Histogram কী?
Histogram হলো এমন একটি Graph যা Numerical Data এর Frequency Distribution দেখায়।\
এখানে:
- x-axis Numerical Value
- y-axis - Frequency
### bins=50 এর অর্থ কী?
Histogram এ Data কে কয়েকটি Group বা Interval এ ভাগ করা হয়।
এই Group গুলোকে Bin বলা হয়।\
এখানে:\
`bins=50`\
অর্থাৎ Data কে 50 টি Interval এ ভাগ করা হয়েছে।
#### Histogram থেকে কী বোঝা যায়?
Histogram দেখে বোঝা যায়:
- কোন Range & Data বেশি
- Distribution Symmetric নাকি না
- Data Skewed কি না
- Multiple Peak আছে কি না
### Age Histogram বিশ্লেষণ
Age Histogram থেকে সাধারণত দেখা যায়:
- মাঝামাঝি বয়সের Passenger বেশি
- খুব কম বয়সী এবং খুব বেশি বয়সী Passenger কম
- Distribution কিছুটা Right Skewed হতে পারে

### KDE Plot তৈরি করা
এখন একই Data এর KDE Plot দেখা যাক।

```python
sns.kdeplot(df['Age'])

plt.title('KDE plot of Age')
plt.xlabel('Ages')
plt.ylabel('probabilty distribution')

plt.show()
```

# KDE Plot কী?
KDE এর পূর্ণরূপ হলো Kernel Density Estimation।\
এটি Histogram এর একটি Smooth Version।\
KDE Plot Data এর Probability Distribution Smooth Curve আকারে দেখায়।
### KDE Plot কেন ব্যবহার করা হয়?
KDE Plot ব্যবহার করা হয় কারণ:
- Distribution আরও Smooth দেখা যায়
- Pattern সহজে বোঝা যায়
- Peak শনাক্ত করা সহজ হয়
Histogram এবং KDE Plot এর পার্থক্য
Histogram:
Bar ব্যবহার করে।
Frequency দেখায়
KDE Plot:
Smooth Curve ব্যবহার করে
Probability Density দেখায়

### Fare Column এর Histogram
এখন Fare Feature বিশ্লেষণ করা যাক।

```python
sns.histplot(df['Fare'],bins=50)

plt.title('Histogram of Fare')
plt.xlabel('Fare')
plt.ylabel('frequency')

plt.show()
```

### Fare Distribution থেকে কী বোঝা যায়?
Fare Histogram এ সাধারণত দেখা যায়:
- অধিকাংশ Passenger কম Fare দিয়েছে
- কিছু Passenger অনেক বেশি Fare দিয়েছে
- Distribution অনেক Right Skewed
### Right Skewed Distribution কী?
যখন বেশিরভাগ Data ছোট Value এর দিকে থাকে কিন্তু কিছু বড় Value\ ডানদিকে Tail তৈরি করে, তখন তাকে Right Skewed Distribution বলে।

### Boxplot কী?
Boxplot হলো এমন একটি Visualization যা Data এর Spread এবং Outlier দেখায়।\
এটি সাধারণত নিচের বিষয়গুলো দেখায়:
- Minimum Value
- Maximum Value
- Median
- Quartile
- Outlier
### Boxplot এর অংশগুলো
একটি Boxplot এ সাধারণত থাকে:
- Box
- Median Line
- Whisker
- Outlier Point
### Median কী?
Median হলো Dataset এর মাঝের Value।\
৫০% Data Median এর নিচে থাকে।
### Quartile কী?
Quartile Data কে চার ভাগে ভাগ করে।
- Q1 -25%
- Q2 50% (Median)
- Q3 - 75%
### Outlier কী?
যেসব Value অন্যান্য Value থেকে অনেক দূরে থাকে সেগুলোকে Outlier বলে।\
উদাহরণ:\
যদি অধিকাংশ Fare হয়:
- 10 থেকে 50

কিন্তু কিছু Fare হয়:
- 300 বা 500

তাহলে সেগুলো Outlier হতে পারে।

### Fare Boxplot বিশ্লেষণ
Fare Boxplot এ সাধারণত দেখা যায়:

- Distribution Right Skewed
- বেশিরভাগ Fare কম Range এ আছে
