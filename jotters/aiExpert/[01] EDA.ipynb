## 1.1 What is EDA?

EDA এর পূর্ণরূপ হলো **Exploratory Data Analysis**।\
বাংলায় একে বলা যায় **ডেটা অনুসন্ধানমূলক বিশ্লেষণ**।

Machine Learning বা Data Science প্রজেক্টে Model তৈরির আগে Dataset সম্পর্কে ভালোভাবে বোঝার জন্য যে বিশ্লেষণ করা হয়, সেটাই EDA।

সহজভাবে বললে, EDA হলো Dataset এর ভিতরে কী আছে, কোন ধরনের তথ্য আছে, কোথায় সমস্যা আছে, কোন Pattern আছে এসব খুঁজে বের করার প্রক্রিয়া।

**EDA** is a systematic process of investigating datasets to understand their structure, patterns, and quality before building machine learning models.

### Why EDA Matters?
অনেক সময় Dataset দেখতে ঠিক মনে হলেও ভিতরে বিভিন্ন সমস্যা থাকতে পারে। যেমন:

- Missing Value থাকতে পারে
- ভুল Data Type থাকতে পারে
- Duplicate Data থাকতে পারে
- Outlier থাকতে পারে
- Irrelevant Feature থাকতে পারে
- Data Imbalance থাকতে পারে

যদি এসব সমস্যা আগে থেকে শনাক্ত না করা হয়, তাহলে Machine Learning Model ভালো Result দিতে পারবে না।\
তাই EDA হলো Data Science Workflow এর সবচেয়ে গুরুত্বপূর্ণ ধাপগুলোর একটি।

There an AI principle known as \
**"Garbage In, Garbage Out" (GIGO)**

Poor quality data leads to unreliable models. EDA helps ensure data quality and reliability before proceeding to model development.

### Key Objectives:

EDA করার সময় সাধারণত নিচের বিষয়গুলো বোঝার চেষ্টা করা হয়:\

**i. Dataset Structure বোঝা**
- কতগুলো Row আছে
- কতগুলো Column আছে
- কোন Column Numerical
- কোন Column Categorical

**ii. Data Quality যাচাই করা**

- Missing Value আছে কি না
- Duplicate আছে কি না
- Invalid Data আছে কি না

**iii. Statistical Understanding তৈরি করা**

- Mean
- Median
- Mode
- Standard Deviation
- Minimum & Maximum Value

**iv. Pattern এবং Relationship খুঁজে বের করা**

- কোন Feature গুরুত্বপূর্ণ
- কোন Feature Target এর সাথে সম্পর্কিত
- কোন Feature অপ্রয়োজনীয়

**v. Visualization এর মাধ্যমে Data বোঝা**

Graph ব্যবহার করে Data আরও সহজে বিশ্লেষণ করা যায়।\

যেমন:

- Histogram
- Boxplot
- Countplot
- Scatterplot




### EDA Process:
1. **Data Loading & Overview** - Shape, columns, data types
2. **Data Cleaning** - Handle missing values, duplicates, inconsistencies
3. **Descriptive Statistics** - Mean, median, std, variance, distribution
4. **Univariate Analysis** - Analyze individual variables
5. **Bivariate & Multivariate Analysis** - Study relationships between variables
6. **Data Visualization** - Use charts and plots to identify patterns
7. **Feature Engineering** - Transform and prepare features for modeling


**Aspect:**
1. Type of data, Count data
2. Data distribution
3. Spotting Missing data
4. Outliers
5. Relationship between features

### Methods:
**1. Analytical View of a Dataset:**
Shape, columns, data types, missing values, duplicates, etc.

**2. Feature Screening**

--> Identifying which features are relevant, redundant, or need transformation.

**3. Statistical Analysis**
Understanding data through statistics — mean, median, std, variance, correlation, etc.

**4. Univariate Analysis**
Analyzing **one variable at a time**:
- **Categorical** → bar chart, frequency count
- **Numerical** → histogram, boxplot, distribution

**5. Bi-variate & Multivariate Anaysis**

**6. Y-data Profiling**
