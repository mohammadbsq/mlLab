### Dataset এর Shape দেখা
Dataset এ কতগুলো Row এবং Column আছে তা দেখার জন্য shape ব্যবহার করা হয়।

```python
df.shape
```

### Total Feature এবং Sample Data বের করা
এখন আলাদাভাবে Total Feature এবং Total Sample Data দেখা যাক।

```python
print(f"Total potential features :{df.shape[1]}")
print(f"Total sample data :{df.shape[0]}")
```

এখানে:\
df.shape[1] Column সংখ্যা দেয়\
df.shape[0] Row সংখ্যা দেয়

### কেন Shape গুরুত্বপূর্ণ?
Dataset এর Shape জানলে আমরা বুঝতে পারি:
- Dataset বড় নাকি ছোট
- কত Feature নিয়ে কাজ করতে হবে

যদি Dataset এ:

### Dataset এর প্রথম কিছু Row দেখা
Dataset এর শুরু থেকে কয়েকটি Row দেখার জন্য head() ব্যবহার করা হয়।

```python
df.head(10)
```

এখানে:
- প্রথম 10 টি Row দেখানো হবে

এই Method ব্যবহার করে আমরা বুঝতে পারি:
- Column Name কী
- Data Type কেমন হতে পারে
- Missing Value আছে কি না
- Numerical নাকি Categorical Data

### কেন head() ব্যবহার করা হয়?
কারণ বড় Dataset পুরোটা একসাথে দেখা সম্ভব নয়।\
তাই শুরুতে কয়েকটি Row দেখে Dataset সম্পর্কে Initial Understanding তৈরি করা হয়।


### Random Sample দেখা
শুধু প্রথম Row দেখলে অনেক সময় Dataset সম্পর্কে ভুল ধারণা হতে পারে।/
তাই Random Sample দেখাও গুরুত্বপূর্ণ।

```python
df.sample(10)
```

এখানে:\
Dataset থেকে Randomly 10 টি Row দেখানো হয়েছে

### sample() কেন গুরুত্বপূর্ণ?
কারণ অনেক সময় Dataset এর শুরু এবং শেষ অংশে Pattern আলাদা হতে পারে।\
Random Sample দেখলে:
- Data Distribution সম্পর্কে ধারণা পাওয়া যায়
- Random Observation যাচাই করা যায়
- Data Entry সমস্যা বোঝা যায়

### head() এবং sample() এর পার্থক্য

head():
- শুরু থেকে Data দেখায়
- Sequential View দেয়

sample():
- Random Data দেখায়
- Better Overall Understanding দেয়
