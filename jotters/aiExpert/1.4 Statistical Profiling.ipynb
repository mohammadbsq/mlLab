EDA এর একটি গুরুত্বপূর্ণ ধাপ হলো Statistical Profiling।\
এই ধাপে Dataset এর বিভিন্ন Statistical Information বিশ্লেষণ করা হয় যাতে Data সম্পর্কে গভীর ধারণা পাওয়া যায়।

সহজভাবে বললে, Statistical Profiling হলো Dataset এর সংখ্যাগত এবং গঠনগত বৈশিষ্ট্য বিশ্লেষণ করার প্রক্রিয়া।

এই ধাপে সাধারণত নিচের বিষয়গুলো দেখা হয়:

- Median

- Maximum Value

### Statistical Summary দেখা
Numerical Column এর Statistical Summary দেখার জন্য describe() ব্যবহার করা হয়।

```python
df.describe()
```

### describe() কী দেখায়?
describe() সাধারণত Numerical Column এর জন্য নিচের Statistical Information দেখায়:
- count
- mean
- std
- min
- 25%
- 50%
- 75%
- max

#### count
count দেখায় কতগুলো Non-null Value আছে।\
উদাহরণ:
যদি Age Column এর count = 714 হয়,\
তাহলে বোঝা যায় বাকি Row গুলোতে Missing Value আছে।
#### mean
mean হলো Average Valuel\
সূত্র:
$\text{Mean} = \frac{\sum x}{n}$

উদাহরণ:\
যদি Passenger Age এর Mean = 29.7 হয়,\
তাহলে Average Passenger Age প্রায় 30 বছর।

#### std (Standard Deviation)
Standard Deviation দেখায় Data কতটা Spread হয়েছে।\
কম Standard Deviation:
- Data কাছাকাছি আছে

বেশি Standard Deviation:
- Data বেশি ছড়িয়ে আছে

#### min এবং max
min = সবচেয়ে ছোট Value\
max = সবচেয়ে বড় Value\
এগুলোর মাধ্যমে Data Range বোঝা যায়।
#### Percentile কী?
describe() এ 25%, 50%, এবং 75% দেখানো হয়। এগুলোকে Percentile বলা হয়।
#### 25% (First Quartile)
২৫% Data এই Value এর নিচে থাকে।
#### 50% (Median) 
৫০% Data এই Value এর নিচে থাকে।\
এটি Median হিসেবেও পরিচিত।
#### 75% (Third Quartile) 
৭৫% Data এই Value এর নিচে থাকে।

### কেন Percentile গুরুত্বপূর্ণ?
Percentile ব্যবহার করে:
- Data Distribution বোঝা যায়
- Outlier শনাক্ত করা যায়
- Skewness সম্পর্কে ধারণা পাওয়া যায়
### Outlier সম্পর্কে ধারণা
যদি max Value অনেক বেশি হয় এবং 75% এর কাছাকাছি না থাকে, তাহলে Outlier থাকার সম্ভাবনা থাকে।\
উদাহরণ:
- 75% Fare = 31
- max Fare = 512
এখানে Fare Column & Outlier থাকতে পারে।
