## 1.5 Univariate Analysis (Categorical Value Analysis)

### Univariate Analysis কী?
Univariate Analysis হলো এমন একটি Analysis যেখানে একবারে শুধুমাত্র একটি Feature বা Column বিশ্লেষণ করা হয়। এখানে আমরা একটি Column এর Distribution, Frequency এবং Pattern বোঝার চেষ্টা করি।

যদি Column টি Categorical হয়, তাহলে সাধারণত দেখা হয়:
- কোন Category কতবার এসেছে
- কোন Category বেশি বা কম
- Percentage Distribution কেমন
- Data Balanced নাকি Imbalanced
### Categorical Data কী?
Categorical Data হলো এমন Data যা বিভিন্ন Category বা Group নির্দেশ করে।\
উদাহরণ:
- Male or Female
- Yes or No
- Survived or Not Survived
- 1st Class or 2nd Class or 3rd Class

Titanic Dataset এ কিছু Categorical Feature হলো:
- Survived
- Pclass
- Sex
- Embarked

### কেন Categorical Analysis গুরুত্বপূর্ণ?
Categorical Analysis এর মাধ্যমে আমরা বুঝতে পারি:
- কোন Category Dominant
- Data Imbalance আছে কি না
- কোন Category Rare
- Feature সম্পর্কে প্রাথমিক Insight
এটি ভবিষ্যতে Model Building এবং Feature Engineering এ সাহায্য করে।

### Survived Column Analysis
প্রথমে Survived Column বিশ্লেষণ করা যাক।

```python
sns.countplot(data = df , x=df['Survived'])
```

### countplot() কী করে?
countplot() প্রতিটি Category কতবার এসেছে তা Graph আকারে দেখায়।\
এখানে:
- x-axis & Category থাকে
- y-axis & Count থাকে

### Survived Column এর অর্থ
Titanic Dataset এ সাধারণত:
- 0 --> Passenger Survive করেনি
- 1 --> Passenger Survive করেছে

### Countplot থেকে কী বোঝা যায়?
Countplot দেখে বোঝা যায়:
- কোন Category বেশি
- Data Balanced নাকি না
- Distribution কেমন

যদি 0 এর Count অনেক বেশি হয়, তাহলে বোঝা যায় বেশি Passenger বাঁচেনি।

### Percentage Calculation
শুধু Count দেখলে সবসময় পরিষ্কার ধারণা পাওয়া যায় না।\
তাই Percentage বের করা গুরুত্বপূর্ণ।

```python
(df['Survived'].value_counts() / len(df) ) * 100
```

### value_counts() কী করে?
value_counts() প্রতিটি Category এর Frequency Count বের করে।\
উদাহরণ:

```python
df['Survived'].value_counts()
```

### Percentage Formula
Percentage বের করার ধারণা:
$\text{Percentage} = \frac{\text{Category Count}}{\text{Total Count}} \times 100$ 

এখানে:
- Category Count = নিদিষ্ট Category এর সংখ্যা
- Total Count = মোট Row সংখ্যা

### Pie Chart তৈরি করা
এখন Pie Chart ব্যবহার করে Survival Distribution দেখা যাক।

```python
survived_count =  df['Survived'].value_counts()  # noqa: F821

survived_label = df['Survived'].unique()

plt.pie(survived_count, labels = survived_label, autopct="%1.1f")
plt.title('Pie chart of survival rate')  # noqa: F821
plt.show()
```

### Pie Chart কেন ব্যবহার করা হয়?
Pie Chart ব্যবহার করা হয়:
- Percentage Distribution বোঝার জন্য
- Category Ratio Visualize করার জন্য\
এখানে প্রতিটি Slice একটি Category নির্দেশ করে।
### autopct="%1.1f" এর কাজ
এটি Pie Chart & Percentage Show করে।\
উদাহরণ:
- 61.6%
- 38.4%
